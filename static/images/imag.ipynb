```python
import os
import json

# Your list of plant folder names
plant_folders = [
    'Aloevera', 'Amla', 'Amruta_Balli', 'Arali', 'Ashoka', 'Ashwagandha',
    'Avacado', 'Bamboo', 'Basale', 'Betel', 'Betel_Nut', 'Brahmi', 'Castor',
    'Curry_Leaf', 'Doddapatre', 'Ekka', 'Ganike', 'Gauva', 'Geranium',
    'Henna', 'Hibiscus', 'Honge', 'Insulin', 'Jasmine', 'Lemon',
    'Lemon_grass', 'Mango', 'Mint', 'Nagadali', 'Neem', 'Nithyapushpa',
    'Nooni', 'Pappaya', 'Pepper', 'Pomegranate', 'Raktachandini', 'Rose',
    'Sapota', 'Tulasi', 'Wood_sorel'
]

base_path = r'D:\CourseProject\ayurvedic\static\images'  # Update if needed

image_data = {}

for folder in plant_folders:
    folder_path = os.path.join(base_path, folder)
    if os.path.exists(folder_path) and os.path.isdir(folder_path):
        images = [f for f in os.listdir(folder_path) if f.lower().endswith('.jpg')]
        images.sort(key=lambda x: int(os.path.splitext(x)[0]) if x.split('.')[0].isdigit() else 0)
        image_data[folder] = images
    else:
        print(f"Warning: Folder not found - {folder_path}")
        image_data[folder] = []

# Save to JSON file
output_path = 'image_list.json'  # Save near your project or where you want
with open(output_path, 'w') as f:
    json.dump(image_data, f, indent=2)

print(f"Image list JSON generated at: {output_path}")
```
